```python
import re

def extract_features(url):
    features = []

    # URL Length
    features.append(len(url))

    # HTTPS
    features.append(1 if url.startswith("https") else 0)

    # Count dots
    features.append(url.count("."))

    # Count hyphen
    features.append(url.count("-"))

    # Count special characters
    features.append(len(re.findall(r'[?=&]', url)))

    # Check IP address
    features.append(1 if re.search(r'\d+\.\d+\.\d+\.\d+', url) else 0)

    # Suspicious keywords
    suspicious_words = ['login','verify','bank','secure','account','update','free','money']

    count = 0
    for word in suspicious_words:
        if word in url.lower():
            count += 1

    features.append(count)

    return features
```

```python
import pandas as pd
from sklearn.ensemble import RandomForestClassifier
import pickle

data = {
    'length':[12,45,60,15,70,80,55,20],
    'https':[1,0,0,1,0,0,0,1],
    'dots':[1,3,4,1,5,4,3,1],
    'hyphen':[0,2,3,0,4,3,2,0],
    'special':[0,2,3,0,5,4,3,0],
    'ip':[0,0,1,0,1,1,0,0],
    'suspicious':[0,2,3,0,4,5,2,0],
    'label':[0,1,1,0,1,1,1,0]
}

df = pd.DataFrame(data)

X = df.drop('label',axis=1)
y = df['label']

model = RandomForestClassifier()
model.fit(X,y)

pickle.dump(model,open('model.pkl','wb'))

print("Model trained successfully!")
```

```python
import pickle
# import extract_features

model = pickle.load(open('model.pkl','rb'))

print("=================================")
print(" AI URL Phishing Detector ")
print("=================================")

url = input("Enter URL: ")

features = extract_features(url)

prediction = model.predict([features])

print("\nAnalyzing...\n")

if prediction[0] == 1:
    print("⚠️ Phishing URL Detected")
else:
    print("✅ Safe URL")
```
